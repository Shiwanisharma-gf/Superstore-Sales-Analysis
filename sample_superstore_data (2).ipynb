```python
# Importing necessary libraries and loading the dataset using 'latin1' encoding to avoid read errors.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
fg = pd.read_csv('SampleSuperstore[1].csv', encoding='latin1')
print(fg)
```

```python
# Checking the data types of each column to ensure they are in the correct format.
fg.dtypes
```

```python
# Finding the number of unique values for each categorical (object) column.
fg.select_dtypes(include='object').nunique()
```

```python
# Displaying the unique values present in the 'Ship Mode' column.
fg['Ship Mode'].unique()
```

```python
# Getting a concise summary of the dataframe, including non-null counts and memory usage.
fg.info()
```

```python
# Checking for any missing (null) values across all columns.
fg.isnull().sum()
```

```python
# Checking the total number of duplicate rows in the dataset.
fg.duplicated().sum()
```

```python
# Removing the duplicate rows to clean the dataset.
fg=fg.drop_duplicates()
```

```python
# Verifying that all duplicate rows have been successfully removed.
fg.duplicated().sum()
```

```python
# Generating descriptive statistics for the numerical columns (mean, min, max, etc.).
fg.describe()
```

```python
# Calculating the total sales for each 'Ship Mode'.
fg.groupby('Ship Mode')['Sales'].sum()
```

```python
# Aggregating data to find the total sales and total quantity count per 'Region'.
fg.groupby('Region').agg({
    'Sales': 'sum',
    'Quantity':'count'
    
})
```

```python
# Displaying the unique customer segments in the dataset.
fg['Segment'].unique()
```

```python
# Calculating the total profit generated by each customer 'Segment'.
fg.groupby('Segment')['Profit'].sum()
```

```python
# Calculating the total discount given to each customer 'Segment'.
fg.groupby('Segment')['Discount'].sum()
```

```python
# Converting the 'Sales' column from float to integer format for cleaner data representation.
fg['Sales']=fg['Sales'].astype(int)
fg
```

```python
# Converting the 'Profit' column from float to integer format.
fg['Profit']=fg['Profit'].astype(int)
fg
```

```python
# Creating a bar chart to visualize total sales across different regions.
region_sales=fg.groupby('Region')['Sales'].sum()
plt.bar(region_sales.index,region_sales.values,label='SALES')
plt.xlabel('Region')
plt.ylabel('Total sales')
plt.legend()
plt.title('Sales by Region')
plt.show()
```

```python
# Creating a horizontal bar chart to show the total quantity sold for each category.
categiores_count=fg.groupby('Category')['Quantity'].count()
plt.figure(figsize=(8,6))
plt.barh(categiores_count.index,categiores_count.values,label='Category count')
plt.xlabel('Category')
plt.ylabel('Quantity')
plt.legend()
plt.title('count by category')
plt.show()
```

```python
# Filtering for positive profit and creating a pie chart for the top 5 profitable states.
state_profit = fg.groupby('State')['Profit'].sum()
positive_profit = state_profit[state_profit > 0].nlargest(5)

plt.figure(figsize=(8, 6))
plt.pie(positive_profit.values, labels=positive_profit.index, autopct='%1.1f%%')
plt.title('Profit by State (Positive Only)')
plt.show()
```

```python
# Creating a combined dashboard with subplots to display regional sales, category
# quantities, and top profitable states toge

# subplots
region_sales=fg.groupby('Region')['Sales'].sum()
categiores_count=fg.groupby('Category')['Quantity'].sum()
state_profit = fg.groupby('State')['Profit'].sum()
fig,(ax1,ax2,ax3)=plt.subplots(1,3,figsize=(14,6))
# bar-1
ax1.bar(region_sales.index,region_sales.values,color='green')
ax1.set_xlabel('Region')
ax1.set_ylabel('Sales')
ax1.set_title('sales by region')
# barh-2
ax2.barh(categiores_count.index,categiores_count.values,label='Category count')
ax2.set_xlabel('category')
ax2.set_ylabel('Quantity')
ax2.legend()
ax2.set_title('quqntity by category')
# pie
ax3.pie(positive_profit.values, labels=positive_profit.index, autopct='%1.1f%%')
ax3.set_title('top 5 satates by profit')
plt.tight_layout()
plt.show()



```
